##### Objective:
To demonstrate incremental data processing in Delta Lake by loading customer data, cleaning the dataset, performing MERGE operations to update and insert records, and validating the final Delta table.

#### Step 1: Load the dataset into a Delta table

```python
master_df=spark.read.option("header","true").option("inferSchema","true").csv("/Volumes/workspace/default/week7_files/customer_master.csv")

display(master_df)
```

#### Check the schema

```python
master_df.printSchema()
```

####Load Master Dataset into Delta Table

```python
master_df.write.format("delta").mode("overwrite").saveAsTable("customer_master_delta")
```

```python
display(master_df.limit(10))
```

#### Step 2: Data Cleaning

```python
delta_df=spark.table("customer_master_delta")
```

```python
print("Rows before cleaning the dataset:", delta_df.count())
```

Checking for null values

```python
from pyspark.sql.functions import col,count,when
delta_df.select([count(when(col(c).isNull(), c)).alias(c)for c in delta_df.columns]).show()
```

Handling null values


```python
delta_df=delta_df.fillna({ "city": "Unknown","status": "Active"})
```

Checking for duplicate records

```python
delta_df.count()
```

```python
delta_df.dropDuplicates().count()
```

Remove duplicates Rows

```python
delta_df=delta_df.dropDuplicates()
```

```python
delta_df=delta_df.na.drop().dropDuplicates(["customer_id"])
```

```python
delta_df.count()
```

Saving the cleaned Delta table

```python
delta_df.write.format("delta").mode("overwrite").saveAsTable("customer_master_cleaned")
```

#### Step 3: Loading the Incremental Dataset

```python
incremental_df=spark.read.option("header","true").option("inferSchema","true").csv("/Volumes/workspace/default/week7_files/customer_incremental.csv")
```

```python
display(incremental_df)
```

```python
incremental_df.count()
```

```python
incremental_df=incremental_df.dropDuplicates(["customer_id"])
```

```python
display(spark.table("customer_incremental_delta"))
```

#### Step 4: Apply MERGE operation to update existing and insert new records.

```python
%sql
MERGE INTO customer_master_cleaned AS target
USING customer_incremental_delta AS source
ON target.customer_id=source.customer_id
WHEN MATCHED THEN
UPDATE SET
target.name=source.name,
target.city=source.city,
target.email=source.email,
target.signup_date=source.signup_date,
target.status=source.status
WHEN NOT MATCHED THEN
INSERT (customer_id,name,city,email,signup_date,status)
VALUES (source.customer_id,source.name,source.city,source.email,source.signup_date,source.status);
```

#### Step 5: Validate results (row count, duplicates).

```python
final_df=spark.table("customer_master_cleaned")

print("Final Row Count:",final_df.count())
```

Checking for duplicate customer_ids

```python
from pyspark.sql.functions import col, count
duplicates=(final_df.groupBy("customer_id").agg(count("*").alias("count")).filter(col("count")>1)
)
display(duplicates)
```

Verifing newly inserted records
